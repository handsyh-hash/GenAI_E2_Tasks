```python
import ollama

stream = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': 'Write a short poem about coding.'}],
    stream=True,
)

for chunk in stream:
    print(chunk['message']['content'], end='', flush=True)
```

```python
import ollama

stream = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': 'Write a short poem about coding.'}],
    stream=True,
    options={
        'temperature': 0.2 # Low temperature for a more structured poem
    },
)

for chunk in stream:
    print(chunk['message']['content'], end='', flush=True)
```

```python
# Zero-Shot Prompts. yes or no question

import ollama

prompt = "Does the planet Mars have a breathable atmosphere for humans? Answer 'Yes' or 'No' only."

response = ollama.chat(
    model='llama3.2', 
    messages=[{'role': 'user', 'content': prompt}],
    options={'temperature': 0}
)

print(response['message']['content'])
```

```python
# Few shot prompts example. 
import ollama

# The 'system' prompt defines the persona and the rule
system_prompt = "You are a support ticket classifier. Categorize the ticket into one of these categories: [Technical, Billing, Account, Feature Request]. Answer with ONLY the category name. No preamble, no explanation."

# The 'few-shot' examples are passed as part of the conversation history
messages = [
    {'role': 'system', 'content': system_prompt},
    
    # Example 1
    {'role': 'user', 'content': "Ticket: 'I can't log into my account, it says password incorrect.'"},
    {'role': 'assistant', 'content': 'Account'},
    
    # Example 2
    {'role': 'user', 'content': "Ticket: 'My screen goes black whenever I click the export button.'"},
    {'role': 'assistant', 'content': 'Technical'},
    
    # Example 3
    {'role': 'user', 'content': "Ticket: 'I was charged twice for my monthly subscription.'"},
    {'role': 'assistant', 'content': 'Billing'},
    
    # Example 4
    {'role': 'user', 'content': "Ticket: 'It would be great if we could have a dark mode for the dashboard.'"},
    {'role': 'assistant', 'content': 'Feature Request'},
    
    # The actual ticket you want to categorize
    {'role': 'user', 'content': "Ticket: 'The API is returning a 500 error when I try to upload a CSV.'"},
]

response = ollama.chat(
    model='llama3.2', 
    messages=messages,
    options={'temperature': 0} # Crucial for consistent categorization
)

print(response['message']['content'])
# Expected Output: Technical
```
